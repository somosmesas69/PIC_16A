## More about strings

```python
s = '0123456789'
# get length
len(s)
```

```python
# concatenate with +
s1 = "abc"
s2 = "def"
s3 = s1 + s2
s3
```

```python
# indexing
s = '0123456789'
print(s[0])
print(s[5])
print(s[-1])
print(s[len(s)-1])
```

```python
# slicing 
# s[start:end:step] does NOT include s[end]
# if there is no step, assume 1.
# if there is no start, assume 0.
# if there is no end, assume len(s).
# very intuitive once you see it a couple of times, s[:x] and s[-x:] have length x.
s = '0123456789'
print(s[1:5:2])
print(s[2:8:3])
print(s[1:5])
print(s[:5])
print(s[3:])
print(s[-3:])
print(s[::-1]) #reverse string
print(s[::-5])
```

More string methods https://docs.python.org/3/library/stdtypes.html#string-methods 

Definitely *don't* sit down and memorize them, but keep this link handy!

```python
# some commonly used string methods
#strip()
s = " spaci ous "
print(s.strip())
```

```python
s = "aa hhalll aa"
print(s.strip('a'))
print(s.strip('a h'))
print(s.strip('ah '))
```

```python
# join
l = ['ba','na', 'na']
print(''.join(l))
print('-'.join(l))
print('/'.join(l))
```

## Lists

```python
L = [0, 1, 2, 3]
```

```python
# lists can have mixed types
L = [0, 1, 'meow', 3.0, '5']
print(type(L))
```

```python
# len, +, indexing, slicing - same as str
print(len(L))
print(L[0], L[-1])
print(L[:4:2])
print(L[2:])
```

```python
# append and extend
L = [0,1,2,3,4]
L.append(5)
print(L)
L.extend([6,7])
print(L)
L.extend([8])
print(L)
```

```python
# item assignment
L[0] = 99
L
```

## Objects

Objects in Python are abstractions of data.

https://docs.python.org/3/reference/datamodel.html#objects-values-and-types

In `x = 0`, then `0` is the object, and `x` is the name (variable name).

An object has 3 things:
- an identity: basically object's address in memory. use `id()`
- a type: use `type()`
- a value

```python
x = 0
print(type(x),id(x))
```

![](object.jpeg)

```python
x = 0
y = x
print(x,y)
print(id(x),id(y))
```

```python
x = [0]
y = x
print(x,y)
print(id(x),id(y))
```

```python
x = 0
y = 0
print(x,y)
print(id(x),id(y))
```

```python
x = [0]
y = [0]
print(x,y)
print(id(x),id(y))
```

### + and append/extend work differently:

```python
x = [0]
y = x
print(id(x),id(y))
```

```python
x.append(1)
print(x, y)
print(id(x),id(y))
```

```python
x.extend([2,3])
print(x, y)
print(id(x),id(y))
```

```python
x = x + [5]
print(x, y)
print(id(x),id(y))

# note that not only are they different from each other, they are also different from the original object!
```

### `is` vs. `==`: https://docs.python.org/3/library/stdtypes.html#truth-value-testing

```python
x = [0]
y = [0]
print(x is y)
print(x == y)
```

```python
x = 0
y = [0]
print(x is y)
print(x == y)
```

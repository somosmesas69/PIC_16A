## Object oriented programming (OOP) 

OOP is a programming paradigm based on the concept of objects.

Basically everything in Python is an object. Objects are bundles of data (properties) and methods (functions or behaviors). A class in Python are similar to classes from C++.

A class defines an abstract set of possible objects sharing certain characteristics. Once you define the class, you can *instantiate* it, which gives you an instance of the class.

Classes serve as templates for bundling data and functionality together. Creating a new class creates a new *type* of object, allowing new *instances* of that type to be made."

https://docs.python.org/3/tutorial/classes.html

```python
# define class
class Cat:
    pass
```

```python
# puppycat is an instance of class Cat
puppcat = Cat()
print(type(puppcat), id(puppcat))
```

## Class variables, instance variables, self

**Class variables are shared among all instances of a class.** They are defined directly in the class body, outside of any methods.

The self prefix is required to refer to local data and functions -- that is, data and functions that are only available within the object.

Additionally, all methods need to take self as their first argument.

```python
class Cat:
    genus = 'Felis'
    species = 'cactus'

    def get_scientific_name(self):
        return self.genus + ' ' + self.species
```

```python
# access class variables
puppycat = Cat()
print(puppycat.genus)
print(puppycat.species)

# call the class function
print(puppycat.get_scientific_name())
print(Cat.get_scientific_name(puppycat))
```

```python
boba = Cat()
print(boba.genus)
print(boba.species)
print(boba.get_scientific_name())
```

**Instance variables are unique to each instance of the class.**

They are usually defined within methods of the class, most commonly within the `__init__` method, which is the constructor method of the class.

```python
class Cat:
    genus = 'Felis'
    species = 'Cactus'
    def __init__(self,size, color):     # the constructor
        self.size = size    
        self.color = color
    
    def describe(self, s):
        print(s, 'color:', self.color,'size:', self.size)
```

```python
# equivalent to Cat.__init__(puppycat, 'medium', 'tortie')
puppycat = Cat('small','brown')
puppycat.describe('Hello i am a cat with')
```

```python
boba = Cat('medium',' white')
boba.describe('Hello i am a cat with')
```

About the interaction between class variables and instance variables:

- If an instance i of a class C has an instance variable v, then i.v will access the instance variable regardless of whether C has a class variable v or not.

- Suppose an instance i of a class C does not have an instance variable v, but C does have a class variable v.
   
  - Referencing i.v will access the class variable;
  - Assigning to i.v will create an instance variable;
  - Deleting i.v will cause an error if there's no instance variable v.

```python
x = 5 # global x 
def f():
    x = 9       # local x
    print(x)

f()
print(x)
```

```python
class C:
    v = 1   # class var 

i = C()
print(type(i))
print(i.v, C.v)
```

```python
# assignment creates an instance variable
i.v = 5
print(i.v, C.v)
```

```python
# assignment changes the class variable
C.v = 9
print(i.v, C.v)
j = C()
print(i.v , j.v)
```

```python
# del deletes the instance variable
del i.v
print(i.v, C.v)
```

```python
del C.v
```

## Magic methods

Magic methods in Python are methods that have two prefix and suffix underscores in their names.

They're special methods that you can define to add "magic" to your classes. 

https://rszalski.github.io/magicmethods/

### An example

```python
# we want vectors that work like [1, 2] + [3, 4] = [4, 6]
print([1,2]+[3,4])
```

```python
class Vector:
    def __init__(self, x, y):       # initialize
        self. x = x
        self.y = y 
    
    def __str__(self):
        return 'Vector(' + str(self.x) + ',' + str(self.y) + ')'
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
```

```python
u = Vector(1,2)
v = Vector(3,4)

print(u)
print(v)
```

```python
w = u + v
print(w)
```

```python
# Wanting to Simulate a Rnd Walk..
import random

# Simulate Steps 
for i in range(10):     # 10 steps
    x = random.choice(['Heads','Tails'])
    print(x)

for i in range(3):
    x = random.choice([-1,1])
    print(x)
```

```python
# Random Walk Func as well as Doc Strings
def random_walk(lower_bound, upper_bound):
    """
    Doc string Notation ..
    Args: 
        ... blah 
    Returns:
        ... blah 
    """
    
    pos = 0             # start at pos 0 
    positions = [0]     # starting at pos 0
    steps = []          # empty list, no steps yet 

    while True:
        x = random.choice([-1,1])
        pos += x            # update pod
        steps.append(x)     # update steps

        # checking bound
        if pos >= upper_bound:
            print('Upper bound reached.')
            break
        elif pos <= lower_bound:
            print('Lower bound reached.')
            break
        
        positions.append(pos)   # update positions
        
    return pos, positions, steps
```

```python
pos, positions, steps = random_walk(100, -100)
```

```python
class ArithmeticDict(dict):
    def __add__(self, other):
        new = self.copy()           # only can do because its dict
        for key, value in other.items():
            if key in new:
                new[key] += value
            else:
                new[key] = value
        return ArithmeticDict(new)
```

```python
pantry = {
    'rice (lbs)': 2,
    'harissa (jars)': 1,
    'onions': 5,
    'lemons': 3
}

shopping_trip = { 
    'rice (lbs)': 1,
    'onions':2,
    'spinach (lbs)':1
}

pantryA = ArithmeticDict(pantry)
shopping_tripeA = ArithmeticDict(shopping_trip)
print(pantryA + shopping_tripeA)
```

Iterators and Iterable Objects

- Has an ***iter*** method that returns an iterator

Iterator
- companion object of an iterable object 
- (oftein) purpose is to enable for loop over iterable
- has a ***next*** method returning the next element of iteration

```python
# Making own iterator 

class boringList:
    def __init__(self,L):
        self.L = L

    def __str__(self):
        return str(self.L)

    def __iter__(self):
        print('boringList.__iter__() called')
        return boringListIterator(self)

class boringListIterator:
    def __init__(self, bL):
        print('boringListIterator object created')
        self.L = bL.L
        self.i = 0      # keep track of curr index
    
    def __next__(self):
        print('boringListIterator.__next__() called')
        if self.i < len(self.L):
            element = self.L[self.i]
            self.i += 1
            return element          # to print
        else:
            print('Stop Iteration')
            raise StopIteration

B = boringList([1,2,3])
for x in B:
    print(x)
```

```python
# Slower motion
BLI = iter(B)           

while True:
    try:
        print(next(BLI))
    except StopIteration:
        break
```

```python
# Merging Class Definitions into one

class boringList:
    def __init__(self,L):
        self.L = L

    def __str__(self):
        return str(self.L)
    
    def __iter__(self):
        self.i = 0          # keep track of position
        return self

    def __next__(self):
        if self.i < len(self.L):
            self.i += 1
            return self.L[self.i-1]
        else:
            raise StopIteration

# See how you you need to have both 
#                                       1) def __iter__(self):
#                                       2) def __next__(self):
```

```python
B = boringList([1,2,3])
for i in B:
    print(i)
```

```python
def silly_generator(data):
    for i in range(0, len(data), 1):
        yield data[i]

stringy = '0123456789'
sil_gen = silly_generator(stringy)

for i in sil_gen:
    print(i)
```

```python
# Now to Generators
# Generators are shortcut to creating Iterators

# (1) Create the Generator
def every_other_generator(data):
    for i in range(0, len(data), 2):
        yield data[i]

# (2) Make Generator Object
data = '0123456789'
Gen = every_other_generator(data)

print(type(Gen))

# Now do normal for loop 

# (3) Call the Generator
for x in Gen:
    print(x)
```

```python
class boringList:
    def __init__(self,L):
        self.L = L

    def __str__(self):
        return str(self.L)

    def bl_generator(self):
        for e in self.L:
            yield e


B = boringList([1,2,3])
for e in B.bl_generator():
    print(e)
```

```python
import random
def while_loop_generator(data):
    iter_num = 0
    while iter_num < 5:
        i = random.choice(range(len(data)))
        iter_num += 1 
        yield data[i]

data = '0123456789'
Gen = while_loop_generator(data)
for i in Gen:
    print(i)
```

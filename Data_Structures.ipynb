## List: mutable data structure

Identity and type of an object are unchangeable once the object is created. but the value can be changed in SOME objects.

'mutable' vs. 'immutable'

```python
# id address does not change when you assign a new value to the existing list
my_list = [1,2,3]
print(id(my_list))
my_list[1] = 9
print(my_list, id(my_list))
```

```python
# list of lists
L = [1,2,[3,4]]
print(L, id(L))
inner = L[2]
print(inner, id(inner))
inner.append(5)
print(inner, id(inner))
print(L, id(L))
```

### List comprehension:

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [] 
for x in fruits:
    if 'a' in x:
        newlist.append(x)

print(newlist)
```

```python
# list comprehension
newlist = [y for y in fruits if 'a' in y]
print(newlist)
```

Syntax of list comprehension:

newlist = expression + for + item in iterable + if + condition

```python
# create a list of even numbers < 20
number_list = [i for i in range(20) if i%2 == 0]
print(number_list)
```

```python
# Multiple for loops and constraints are allowed
number_list1 = [i for i in range(100) if i%2==0 if i%5==0]
print(number_list1)
```

```python
number_list2 = [i for i in range(100) if i%5==0 if i%2==0]
print(number_list2)
```

## Tuples: immutable data structure

```python
# lists are mutable
L = [1,2,3]
print(L, id(L))
L[0] = 9
print(L, id(L))
```

```python
# strings are immutable
s = '123'
print(s,id(s))
print(s[0])
# s[0] = 9 # error
# print(s)
```

```python
# tuples are immutable 
t = (1,2,3)
print(t, id(t), type(t))
# t[0] = 9 #error
# print(t)
```

```python
# slice a tuple
t = (1,4,7, [2,3], 'abc', '1')
print(t[:2])
print(t[3:])
print(t[::-1])
```

```python
# Verify that tuple is immutable
t1 = (1,2,3,4,5)
t2 = t1
print(id(t1), id(t2))
t2 = t2 + (9,10)
print(t1,t2)
print(id(t1), id(t2))
```

```python
# lists are mutable
l1 = [1,2,3,4,5]
l2 = l1
print(id(l1), id(l2))
l2.extend([9,10])
print(l1,l2)
print(id(l1), id(l2))
```

```python
# tuple with one element
t = (1)
print(t, type(t))       # seen as an int
t = (1,)
print(t, type(t))
```

**Tuple with a list is annoying because list is mutable and it is possible to change list and reflect on tuple.**

```python
my_tuple = (1,2,3,'ucla',[5,6,7])
print(type(my_tuple), id(my_tuple))
L = my_tuple[4]
print(L, id(L))
L[0] = 9
print(L, id(L))
print(my_tuple, id(my_tuple))
```

# sets
A set is an unordered collection with no duplicate elements.

```python
# define a set
S = set()
print(S, type(S), id(S))
S.add(8)
print(S, id(S))
```

```python
# Set is also a mutable type
s1 = set()
s2 = s1
print(id(s1), id(s2))
s1.add(10)
s2.add(8)
print(s1,s2)
print(id(s1), id(s2))
```

```python
# remove one element from the existing set
s = {1,2,3,4,5}
print(s,type(s),id(s))
s.remove(5)
print(s, id(s))
```

```python
# no duplicate elements
s = {1,2,2,2,3,4,5,5,8}
print(s, type(s))
```

```python
# set is not subscriptable
s = {1,2,3,4,5}
# print(s[0]) # error
```

```python
# check element in given set or not
s = {1,2,3,4,5}
print(3 in s)
print(8 in s)
```

There is no order in set:
I define two sets S1 and S2 with the same element, but I type them in a different order when I define S1 and S2. But they are the same because S1==S2 returns True.

```python
s1 = {1,2,3,4,5}
s2 = {2,1,5,4,3}
s3 = {1,2,3,4,5}

print(s1 == s2)
print(s1 is s3)
print(id(s1), id(s2), id(s3))
```

```python
# The elements of a set must be immutable
```

```python
# However, we can use a frozenset.
s = {frozenset({1,2,3}), frozenset({4,5})}
print(s)
print(type(frozenset({1,2,3})))
```

```python
# another way to create a new set
s = {1,2,3,4,5}
print(type(s))

# however, you cannot define an empty list using S={}, because this creates a new dictionary
s = {}
print(type(s))
```

## Dictionaries

A dictionary is a set of (key: value) pairs with the requirement that the keys are unique (within one dictionary).

Keys can be anything which is deeply immutable. "Deeply immutable" is a terminology I just made up, so let's clarify...

Strings and numbers can always be keys. Tuples can be used as keys if they contain only strings and numbers.

Tuples can sometimes be used as keys if they contain tuples...

If a tuple contains any mutable object either directly or indirectly, it cannot be used as a key; otherwise, it's fine as a key.

You can’t use lists as keys, since lists can be modified in place using index assignments, slice assignments, or methods like append().

```python
# create a new dict
d = {}
print(type(d))
d = dict()
print(type(d))
```

```python
d1 = {'one':1, 'two':2, 'three':3}
print(d1, type(d1), id(d1))
d2 = dict(one=1, two=2, three=3)
print(d2, type(d2), id(d2))
d3 = dict({'one':1, 'two':2, 'three':3})
print(d3, type(d3), id(d3))
d4 = dict([('one',1),('two',2),('three',3)])
print(d4, type(d4), id(d4))
print(d1 == d2 == d3 == d4)
```

```python
d = {'key1': 'value1', 'key2':'value2', 'key3':'value3'}
print(d['key2'])
```

```python
# keys must be unique, values can be same
d = {'key1': 'value1', 'key2':'value2', 'key3':'value2'}
print(d['key2'])
```

```python
#delete keyword
d = {'key1': 'value1', 'key2':'value2', 'key3':'value3'}
del d['key2']
print(d)
```

```python
print(d.keys())
print(d.values())
print(d.items())
```

```python
# update
d = {'key1': 'value1', 'key3':'value3'}
print(d, id(d))
d.update({'key1': 'value4'})
print(d, id(d))
```

```python
# for loop for dictionary
for i in d:
    print(i)
```

```python
# you only iterate keys. But if you know keys, you know values attached to keys, see the following example:
for i in d:
    print(i)
    print(d[i])
```

```python
# dictionary comprehension
d = {i: i*i for i in range(8) if i%2==1}
print(d)

# equivalent to a for loop"
for i in range(8):
    if i%2 == 1:
        d[i] = i*i

print(d)
```

```python
# dictionary with a list
d = {'key1': [1,2,3,4], 'key2': [5,6], 'key3': 7}
print(d, id(d))
# print(d[0]) # error
L = d['key1']
print(L)
L.append(9)
print(L)
print(d, id(d))
```

You can learn more about dictionaries here: https://docs.python.org/3.7/library/stdtypes.html#dict

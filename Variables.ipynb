# Variables in Python

Jupyter notebook files have extensions of .ipynb. 
Python script files have extensions of .py.

Python comments start with `#`.

Unlike in C++, you don't need to specify beforehand what data type the variable will be.

- int, float, bool, str

- `print(x)`: prints the value of object named `x`.
- `type(x)`: returns the data type of object named `x`.

```python
# int
# don't write int x = 5
x = 5
```

```python
print(x)
print(type(x))
```

```python
# float
y = 5.0
print(y, type(y))
```

```python
# bool
flag = True
print(flag)
print(type(flag))
```

```python
flag = False
print(flag)
```

```python
# str
# Strings in Python are surrounded by either single quotation marks, or double quotation marks.
# Python doesn't have a separate char class.
s = 'puppy'
print(s, type(s))
s1 = "cat"
print(s1, type(s1))
```

## Variable naming in Python:
Rules of Variable naming:
- It cannot be a keyword: https://docs.python.org/3/reference/lexical_analysis.html#keywords
- It can only contain A-z, 0-9, _
- case-sensitive, e.g. X is not the same as x
- cannot start with a number, e.g. can only be x1, not 1x

Variable naming suggestion: concisely descriptive

Two common conventions of variable naming: 
- snake_case, e.g. color_of_hat = 'yellow'
- camelCase, e.g. colorOfHat = 'yellow'

```python
hat_color = 'yello'
```

```python
color_of_hat = 'red'
```

```python
colorOfHat = 'yellow'
```

## Arithmetic

```python
print(5 + 9)
print(9 - 5)
print(9 * 5)
print(9 / 5, type(9 / 5))
```

```python
# integer division
print(9 // 5, type(9 // 5))
```

```python
print(5 % 2)
print(5 ** 2)
```

```python
# bitwise XOR
print(5 ^ 2)

# 5 = 2**2 * 1 + 2**1 * 0 + 2**0 * 1
# 2 = 2**2 * 0 + 2**1 * 1 + 2**0 * 0
# 7 = 2**2 * 1 + 2**1 * 1 + 2**0 * 1
# 101
# 010
# 111
```

101 (binary representation of 5)
010 (binary representation of 2)
111 (binary representation of 7)

1 XOR 0 => 1
0 XOR 1 => 1
1 XOR 1 => 1
0 XOR 0 => 0

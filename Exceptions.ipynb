## Raising Exceptions

Runtime errors raise exceptions. Exceptions prevent code from failing silently.

Built-in exceptions: https://docs.python.org/3/library/exceptions.html#bltin-exceptions

Some of the most common ones: https://www.programiz.com/python-programming/exceptions


```python
def halve(x):
    return x/2
```

```python
halve(5)
```

```python
# but this function isn't designed to work with strings in the first place
halve('five')
```

While this basically works, the design of the function doesn't fully reflect our intention. The problem isn't that the function tried to divide a string by an integer, the problem is that this function *isn't designed to work with strings in the first place.* We can reflect our intention by **raising** a more helpful exception. 

`raise`: when you `raise` an exception, it immediately halts the execution of the code.

```python
def halve(x):
    if type(x) not in [float, int]:
        raise TypeError('This fun is designed to work with float or int')
    return x/2
```

```python
halve('five')
```

## Handling Exceptions

In some cases, you might want to **handle** an exception: that is, acknowledge it while allowing your code to continue executing. This can be done via `try`-`except` blocks. The `try` clause states the "expected" behavior, while the `except` clause states what should happen in case of a specified error. Like raising an exception, the `except` clause can inform your user of what has happened. Unlike raising an exception, the `except`  clause can also perform additional computations. 

```python
# If no error occurs while executing the try clause, then the except clause is not executed.
try:
    print('No errors here.')
    1/0
except:
    print('There was an error')
```

```python
# If an error occurs while executing the try clause,
# execution of the try clause stops and the except clause is executed.
try:
    print('before 1/0')
    0 + '1'
    print('After 1/0')
except:
    print('there was an error')
```

A `try`-`except` statement like the one above accepts any exception.

We can be more precise about the errors which we *handle*
(and it is encouraged that you try to be precise).

```python
# a `ZeroDivisionError` error occurs while executing the `try` clause. 
try:
    print('before 1/0')
    1/0
    print('after 1/0')
except ZeroDivisionError:
    print('there was an error')
```

```python
# a `TypeError` error occurs while executing the `try` clause. 
#The `except` statement only handles `ZeroDivisionError`s, 
# so the exception is *raised* without the `try`-`except` statement: execution halts.
try:
    print('Before 1/0')
    0 + '0'
    print('After 1/0')
except ZeroDivisionError:
    print('there was an error')
```

We can handle multiple types of exception together.

```python
try:
    print('before 1/0')
    0 + '0'
    print('after 1/0')
except (ZeroDivisionError, TypeError):
    print('tjere was an error')
```

We can also handle different exceptions separately using multiple except clauses.

```python
try:
    print('before 1/0')
    0 + '0'
    1/0
except ZeroDivisionError:
    print('There was a zero division error')
except TypeError:
    print('There was a type error')
```

Comment: When error occurs in try part, then the code stops immediately and enters the except part. In the following example, code runs `1/0` and ZeroDivisioError occurs, then code breaks immediately and enters except statement. In other words, the code does not see `0+"0"`.

```python
try:
    print('before 1/0')
    0 + '0'
    1/0
    print('After 1/0')
except ZeroDivisionError:
    print('there was a zero division error')
except TypeError:
    print('there wss a typeerror')
```
